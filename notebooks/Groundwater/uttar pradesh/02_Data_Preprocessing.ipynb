```python
import pandas as pd 
import numpy as np 

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv('../../../data/raw/up_groundwater_telemetry.csv')
```

```python
df.isnull().sum()
```

```python
df.drop(columns='RL_MSL', inplace=True)
df.shape
```

```python
columns_to_drop = [
    'SlNo', 
    'Agency', 
    'State LGD Code', 
    'River', 
    'Basin', 
    'Tributary', 
    'Subtributary', 
    'SubSubtributary', 
    'Local River'
] 

df.drop(columns= columns_to_drop, inplace=True) 
df.shape
```

```python
df.drop(columns='Block', inplace=True) 
df.shape
```

```python
df.drop(columns='Village', inplace=True)
df.shape
```

```python
# now we will Convert Date Column:  

df['Data Acquisition Time'] = pd.to_datetime(df['Data Acquisition Time'], format= '%d-%m-%Y %H:%M') 

df.info()
```

```python
df.drop(columns='Tehsil', inplace=True) 
df.shape
```

```python
df.to_csv('../../../data/interim/up_cleaned_groundwater.csv', index=False)
```
