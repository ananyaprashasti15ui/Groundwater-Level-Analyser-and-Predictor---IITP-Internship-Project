```python
import pandas as pd 
import numpy as np 

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv('../../../data/raw/punjab_groundwater_telemetry.csv')
```

```python
df.isnull().sum()
```

```python
df.drop(columns='RL_MSL', inplace=True)
df.shape
```

There are also many unnecessary columns such as:

"SlNo",

"Agency",

"State LGD Code",

"River",

"Basin",

"Tributary",

"Subtributary",

"SubSubtributary",

"Local River"

so we will drop these also

```python
columns_to_drop = [
    'SlNo', 
    'Agency', 
    'State LGD Code', 
    'River', 
    'Basin', 
    'Tributary', 
    'Subtributary', 
    'SubSubtributary', 
    'Local River'
] 

df.drop(columns= columns_to_drop, inplace=True) 
df.shape
```

now we will inspect two more columns i.e. Block and Village since these also carry mostly "-" value

```python
df['Block'].value_counts(dropna=False)
```

so we will not drop Block column as it has only 24% '-' values, therefore we will keep it and replace '-' with null value

but as we can see there are many inconsistent spellings of block such as: 

MOGA-I / MOGA - I

SIDHWANBET / SIDHWAN BET

GHANOUR / GHANAUR

DHARAMKOT / DHARMKOT

HARSHA CHINA / HARSA CHHINA

SARDULGARH / SARDUL GARH

```python
#so first we will replace '-' with NaN values 

df["Block"] = df["Block"].replace("-", np.nan)

df['Block'].isnull().sum()
```

```python
# now we will create a Mapping Dictionary: 

block_mapping = {
    "MOGA - I": "MOGA-I",
    "SIDHWAN BET": "SIDHWANBET",
    "GHANAUR": "GHANOUR",
    "DHARMKOT": "DHARAMKOT",
    "HARSA CHHINA": "HARSHA CHINA",
    "SARDUL GARH": "SARDULGARH",
    "GAHLL KURD": "GHALL KHURD",
    "LUDHIANA - I": "LUDHIANA-I",
    "LUDHIANA - II": "LUDHIANA-II",
    "LUDHIANA -II / MANGAT": "MANGAT", 
    "JALLANDHAR-EAST": "JALANDHAR-EAST",
    "JALLANDHAR-WEST": "JALANDHAR-WEST",
    "DASUA": "DASUYA",
    "HARSHA CHINA": "HARSA CHHINA",
    "QUADIAN": "QADIAN"
}

df['Block'] = df['Block'].replace(block_mapping)
```

```python
df['Block'].value_counts(dropna=False)
```

```python
sorted(df["Block"].dropna().unique())
```

```python
#now we will check same for village: 

df['Village'].value_counts(dropna=False)
```

```python
#so we will drop this as it doesn't contain much information 

df.drop(columns= 'Village', inplace=True)
```

```python
# now we will Convert Date Column: 

df['Data Acquisition Time'] = pd.to_datetime(df['Data Acquisition Time'], format= '%d-%m-%Y %H:%M') 

df.info()
```

```python
#now we will check Tehsil: 

df['Tehsil'].value_counts(dropna=False)
```

The Tehsil column contains approximately 87% missing values (including both NaN and "-" placeholders). Since only about 13% of the records contain valid tehsil information, the feature provides very limited coverage and would require extensive imputation. Additionally, other geographical features such as District, Block, Latitude, and Longitude already capture location information more consistently. Therefore, we will drop Tehsil column

```python
df.drop(columns= 'Tehsil', inplace=True) 
df.shape
```
